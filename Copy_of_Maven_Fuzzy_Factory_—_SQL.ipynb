```python
import pandas as pd
import sqlite3
import os
import zipfile # Needed for zip extraction

conn = sqlite3.connect("maven_fuzzy_factory.db")

data_dir = 'maven_data'
zip_file_name = 'Maven+Fuzzy+Factory.zip' # Assuming this is the name of the uploaded zip file

# Ensure the data directory exists and contains files
if not os.path.exists(data_dir) or not os.listdir(data_dir):
    print(f"Directory '{data_dir}' is missing or empty. Attempting to extract '{zip_file_name}'.")
    if os.path.exists(zip_file_name):
        try:
            os.makedirs(data_dir, exist_ok=True) # Ensure directory exists before extraction
            with zipfile.ZipFile(zip_file_name, 'r') as zip_ref:
                zip_ref.extractall(data_dir)
            print(f"Successfully extracted '{zip_file_name}' to '{data_dir}'.")
        except Exception as e:
            print(f"Error extracting zip file: {e}")
            print("Please ensure the zip file is uploaded and valid.")
    else:
        print(f"Error: Zip file '{zip_file_name}' not found. Please upload the data zip file using cell KtFbPyP6moj7 first.")

# If directory exists and has content (either initially or after extraction attempt), proceed to load tables
if os.path.exists(data_dir) and os.listdir(data_dir):
    for file in os.listdir(data_dir):
        if file.endswith('.csv') and file != 'maven_fuzzy_factory_data_dictionary.csv':
            table_name = file.replace('.csv', '')
            df = pd.read_csv(os.path.join(data_dir, file))
            df.to_sql(table_name, conn, if_exists='replace', index=False)
    print("Database connection established and tables re-loaded.")
else:
    print("Could not establish database connection with tables as data directory is still not ready.")
```

```python
import pandas as pd
import sqlite3
import os
import zipfile

# Check whether maven_data already exists
if not os.path.exists("maven_data"):

    print("📁 maven_data folder not found.")
    print("Please select Maven+Fuzzy+Factory.zip")

    from google.colab import files
    uploaded = files.upload()

    # Find the uploaded ZIP file
    zip_file = [f for f in uploaded.keys() if f.endswith(".zip")][0]

    # Extract it
    with zipfile.ZipFile(zip_file, "r") as zip_ref:
        zip_ref.extractall("maven_data")

    print("✅ ZIP extracted successfully!")

# Connect to SQLite
conn = sqlite3.connect("maven_fuzzy_factory.db")

# Load CSV files into SQLite
for file in os.listdir("maven_data"):

    if file.endswith(".csv") and file != "maven_fuzzy_factory_data_dictionary.csv":

        df = pd.read_csv("maven_data/" + file)

        table_name = file.replace(".csv", "")

        df.to_sql(
            table_name,
            conn,
            if_exists="replace",
            index=False
        )

        print(f"✅ {table_name} loaded — {len(df)} rows")

print("\n🎉 DATABASE READY!")
```
